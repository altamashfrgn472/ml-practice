```python
class knn:
    def __init__(self, k = 5):
        self.n_neighbors = k
        self.X_train = None
        self.y_train = None

    def fit(self, X_train, y_train):
        self.X_train = X_train
        self.y_train = y_train

    def predict(self, X_test):
        y_pred= []
        
        for i in X_test:
            # calculate distance with each training point
            distances = []
            for j in self.X_train:
                distances.append(self.calculate_distance(i, j))
                
            # print(sorted(list(enumerate(distances)), key = lambda x : x[1])[0: self.n_neighbors])
            distances = enumerate(distances)
            distances = list(distances) # it gives the index number, based on current situation(now it correspondence to trainin data in sequence)
            distances = sorted(distances, key = lambda x : x[1])[0 : self.n_neighbors]
            n_neighbors = distances

            label = self.majority_count(n_neighbors)
            y_pred.append(label)
        return np.array(y_pred)

    def calculate_distance(self, point_A, point_B):
        
        return np.linalg.norm(point_A - point_B)

    def majority_count(self, neighbors):
        votes = []
        for i in neighbors:
            votes.append(self.y_train[i[0]])

        votes = Counter(votes)
        return votes.most_common()[0][0]
        
```

```python
a = knn(k = 5)
a.fit(X_train,y_train)
```
